Q1 Write a program to print the first N terms of the fibonacci Series? i.e. 0,1,1,2,3,5,8....

```python
n=int(input("Enter the number of terms:"))
a,b=0,1
c=a+b
print(a)
print(b)
for i in range(n-2):#0,1 count as first two terms
    print(c)
    a,b=b,c
    c=a+b
print("---END---")
```

Q2 Check if a givien number is prime

```python
n=int(input("Enter a number:"))
if n<=1:
    print(f"{n} is not a prime number")
else:
    for i in range(2,n):
        if n%i==0:
            print(f"{n} is not a prime number")
            break
    else:
        print(f"{n} is a prime number")
```

Q4 Write a guessing game where the user has to guess a secret number. After every guess the program tells the user whether their number was too large or too small. At the end the number of tries needed should be printed. It counts only as one try if they input the same number multiple times consecutively.

```python
import random
secret_number=random.randint(1,100)
gussed=[]
while True:
    guess=int(input("Enter your guess (1-100):"))
    gussed.append(guess)
    if guess<secret_number:
        print(f"{guess} is too low. Try again.")
    elif guess>secret_number:
        print(f"{guess} is too high. Try again.")
    else:
        print(f"You won! The number was {secret_number}.")
        print(f"You took {len(gussed)} attempts. Your guesses were: {gussed}")
        break
```

Q11 Find the Least Common Multiple of A and b

```python
a=int(input("Enter the first number:"))
b=int(input("Enter the second number:"))
if a>b:
    large=a
    small=b
else:
    large=b
    small=a
lcm=1
n=1
while lcm<large*small:
    lcm=large*n
    if (large*n)%small==0:
        break
    n+=1
print(f"The LCM of {a} and {b} is {lcm}")
```

Q12 Find the greatest common divisor of a and b

```python
a=int(input("Enter the first number:"))
b=int(input("Enter the second number:"))
if a>b:
    large=a
    small=b
else:
    large=b
    small=a
hcf=1
for i in range(small,0,-1):
    if large%i==0 and small%i==0:
        hcf=i
        break
print(f"The HCF of {a} and {b} is {hcf}")
```

Q14 Check if the Number N is a palaindrome. Eg- 434 reversed is 434, 256 reversed is 652 so not a palaindrome

```python
n=str(input("Enter the number:"))
list_n=[]
for i in n:
    list_n.append(int(i))
list_n.reverse()
num=""
for i in list_n:
    num+=str(i)
num=int(num)
if num==int(n):
    print(f"{n} is a palindrome number")
else:
    print(f"{n} is not a palindrome number")
```

Q15 Check if a string is a palindrome

```python
n=str(input("Enter the number:")).lower()
list_n=[]
for i in n:
    list_n.append(str(i))
list_n.reverse()
word=""
for i in list_n:
    word+=str(i)
if word==n:
    print(f"{n} is a palindrome")
else:
    print(f"{n} is not a palindrome")
```

Q17 Check if A and B are coprime

```python
a=int(input("Enter the first number:"))
b=int(input("Enter the second number:"))
flag=0
if a>b:
    large=a
    small=b
else:
    large=b
    small=a
for i in range(2,small+1):
    if large%i==0 and small%i==0:
        flag=1
if flag==1:
    print(f"{a} and {b} are not co-prime numbers")
else:
    print(f"{a} and {b} are co-prime numbers")
```

Q20 for a givien string find the number of duplicate characters

```python
word=str(input("Enter the number:"))
n=word.lower()
count={}
for i in n:
    if i in count:
        count[i]+=1
    else:
        count[i]=1
print("Word:",n)
print(count)
```
